# DATA CLEANING

```python
swiggy_clean = swiggy.copy()
```

```python
swiggy_clean.columns = ['restaurant_id', 'area', 'city', 
                         'restaurant_name', 'price', 
                         'avg_rating', 'total_ratings',
                         'food_type', 'address', 
                         'delivery_time_min']
```

```python
df=swiggy_clean.columns = ['restaurant_id', 'area', 'city', 
                         'restaurant_name', 'price', 
                         'avg_rating', 'total_ratings',
                         'food_type', 'address', 
                         'delivery_time_min']
```

```python
df
```

```python
# Price outliers remove karo
swiggy_clean = swiggy_clean[swiggy_clean['price'] > 0]
swiggy_clean = swiggy_clean[swiggy_clean['price'] <= 5000]
```

```python
# Rating valid range
swiggy_clean = swiggy_clean[swiggy_clean['avg_rating'] >= 1]
swiggy_clean = swiggy_clean[swiggy_clean['avg_rating'] <= 5]
```

```python
# Delivery time valid
swiggy_clean = swiggy_clean[swiggy_clean['delivery_time_min'] > 0]
swiggy_clean = swiggy_clean[swiggy_clean['delivery_time_min'] <= 120]
```

```python
# City trim 
swiggy_clean['city'] = swiggy_clean['city'].str.strip()

print(f" Swiggy cleaned: {swiggy_clean.shape}")
```

```python
#DELIVERY CLEANING 
print("\nCleaning Delivery dataset...")

delivery_clean = delivery.copy()
```

```python
delivery_clean.replace("NaN", np.nan, inplace=True)
```

```python
# Weather — "conditions Sunny" → "Sunny"
delivery_clean['Weatherconditions'] = (
    delivery_clean['Weatherconditions']
    .str.replace('conditions ', '', regex=False)
    .str.strip()
)
```

```python
# Time taken — "(min) 24" → 24
delivery_clean['Time_taken(min)'] = (
    delivery_clean['Time_taken(min)']
    .str.replace('(min) ', '', regex=False)
    .str.strip()
    .astype(float)
)
```

```python
#Age or Ratings — string to number
delivery_clean['Delivery_person_Age'] = (
    pd.to_numeric(delivery_clean['Delivery_person_Age'], errors='coerce')
)
delivery_clean['Delivery_person_Ratings'] = (
    pd.to_numeric(delivery_clean['Delivery_person_Ratings'], errors='coerce')
)
```

```python
# ID trim 
delivery_clean['ID'] = delivery_clean['ID'].str.strip()
delivery_clean['City'] = delivery_clean['City'].str.strip()
delivery_clean['Festival'] = delivery_clean['Festival'].str.strip()
delivery_clean['Road_traffic_density'] = (
    delivery_clean['Road_traffic_density'].str.strip()
)
delivery_clean['Type_of_order'] = (
    delivery_clean['Type_of_order'].str.strip()
)
delivery_clean['Type_of_vehicle'] = (
    delivery_clean['Type_of_vehicle'].str.strip()
)
```

```python
# NaN rows drop
delivery_clean.dropna(subset=['Delivery_person_Age', 
                               'Delivery_person_Ratings',
                               'Time_taken(min)'], inplace=True)
```

```python
# Date column
delivery_clean['Order_Date'] = pd.to_datetime(
    delivery_clean['Order_Date'], dayfirst=True, errors='coerce'
)
```

```python
print(f" Delivery cleaned: {delivery_clean.shape}")
```

```python
# VERIFICATION 
print("\n AFTER CLEANING")
print("Swiggy nulls:", swiggy_clean.isnull().sum().sum())
print("Delivery nulls:", delivery_clean.isnull().sum().sum())
print("\nDelivery Cities:", delivery_clean['City'].unique())
print("Weather types:", delivery_clean['Weatherconditions'].unique())
print("Traffic types:", delivery_clean['Road_traffic_density'].unique())
print("\nDelivery time sample:", 
      delivery_clean['Time_taken(min)'].head())
print("\n Both datasets CLEAN and ready!")
```

# Statistics

```python
from scipy import stats

print("=" * 50)
print("   INDIA FOOD DELIVERY — STATISTICS")
print("=" * 50)
```

# ANALYSIS 1 — Average Delivery Time

```python
avg_time = delivery_clean['Time_taken(min)'].mean()
med_time = delivery_clean['Time_taken(min)'].median()
std_time = delivery_clean['Time_taken(min)'].std()

print(f"\n DELIVERY TIME ANALYSIS")
print(f"Average  : {avg_time:.1f} minutes")
print(f"Median   : {med_time:.1f} minutes")
print(f"Std Dev  : {std_time:.1f} minutes")
print(f"Min Time : {delivery_clean['Time_taken(min)'].min()} min")
print(f"Max Time : {delivery_clean['Time_taken(min)'].max()} min")
```

# ANALYSIS 2 — Weather Impact

```python
print(f"\n  WEATHER vs DELIVERY TIME")
print("-" * 35)

# "NaN" weather rows hata do
weather_data = delivery_clean[
    delivery_clean['Weatherconditions'] != 'NaN'
]
weather_analysis = (
    weather_data
    .groupby('Weatherconditions')['Time_taken(min)']
    .mean()
    .sort_values(ascending=False)
    .round(1)
)
print(weather_analysis)

# Sabse slow weather
worst_weather = weather_analysis.index[0]
worst_time = weather_analysis.iloc[0]
print(f"\n⚠️  Slowest: {worst_weather} = {worst_time} min avg")
```

# ANALYSIS 3 — Traffic Impact

```python
print(f"\n TRAFFIC vs DELIVERY TIME")
print("-" * 35)

traffic_data = delivery_clean[
    delivery_clean['Road_traffic_density'] != 'NaN'
]

traffic_analysis = (
    traffic_data
    .groupby('Road_traffic_density')['Time_taken(min)']
    .mean()
    .sort_values(ascending=False)
    .round(1)
)
print(traffic_analysis)
```

# ANALYSIS 4 — Festival Impact

```python
print(f"\n FESTIVAL vs DELIVERY TIME")
print("-" * 35)

festival_yes = delivery_clean[
    delivery_clean['Festival'] == 'Yes'
]['Time_taken(min)'].mean()

festival_no = delivery_clean[
    delivery_clean['Festival'] == 'No'
]['Time_taken(min)'].mean()

diff = festival_yes - festival_no
print(f"Festival days   : {festival_yes:.1f} min")
print(f"Normal days     : {festival_no:.1f} min")
print(f"Extra time      : +{diff:.1f} min on festivals")
```

# ANALYSIS 5 — Correlation

```python
print(f"\n CORRELATION ANALYSIS")
print("-" * 35)

# Rating vs Delivery Time
corr1, p1 = stats.pearsonr(
    delivery_clean['Delivery_person_Ratings'].dropna(),
    delivery_clean['Time_taken(min)'].dropna()
)
print(f"Rating vs Time : {corr1:.3f}")
if corr1 < 0:
    print("  → Higher rating = Faster delivery ")
else:
    print("  → Higher rating = Slower delivery ")
# Age vs Delivery Time
corr2, p2 = stats.pearsonr(
    delivery_clean['Delivery_person_Age'].dropna(),
    delivery_clean['Time_taken(min)'].dropna()
)
print(f"Age vs Time    : {corr2:.3f}")
if corr2 > 0:
    print("  → Older person = Slower delivery")
else:
    print("  → Older person = Faster delivery")
```

# ANALYSIS 6 — City Type

```python
print(f"\n  CITY TYPE vs DELIVERY TIME")
print("-" * 35)

city_data = delivery_clean[
    delivery_clean['City'] != 'NaN'
]

city_analysis = (
    city_data
    .groupby('City')['Time_taken(min)']
    .agg(['mean', 'count'])
    .round(1)
)
city_analysis.columns = ['Avg_Time', 'Orders']
print(city_analysis)
```

# ANALYSIS 7 — Swiggy Price Analysis

```python
print(f"\n SWIGGY — PRICE vs RATING")
print("-" * 35)

corr3, p3 = stats.pearsonr(
    swiggy_clean['price'],
    swiggy_clean['avg_rating']
)
print(f"Correlation: {corr3:.3f}")
if corr3 > 0.3:
    print("  → Costly restaurants = Better rated ")
elif corr3 < -0.3:
    print("  → Costly restaurants = Worse rated ")
else:
    print("  → Price aur Rating mein weak relation hai")

# Top cities by restaurant count
print(f"\n TOP CITIES — Restaurant Count")
print(swiggy_clean['city'].value_counts().head(5))

print("\nStatistics Analysis Complete!")
```
